```python
# Full descriptions on variables for ease of understanding. (Not common practice for such long variables)

# Example Question: Aaron Invests £5,000 @ 4% annual interest rate for 3 years, calculate the total amount at year 3.


p = 5000 # initial principal investment amount
r = 0.04 # interest rate
n = 1 # number of times interest is applied per time period
t = 3 # number of periods elapsed


compound_interest = p * (1 + r)**(n * t)

#Display Result
print(f"Final amount after {t} years: £{compound_interest:.2f}")
```

### Compound Interest Formula

$$
A = P \times \left(1 + \frac{r}{n}\right)^{nt}
$$
